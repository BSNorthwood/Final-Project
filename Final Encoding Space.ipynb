In this space we will be encoding the cleaned dataset.
We wil first import libraries that we will use.

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
```

We must load in our .csv file

```python
df = pd.read_csv("Final_Project_dataset_cleaned.csv")
```

We need to start the label encoder

```python
encoder = LabelEncoder()
```

We will encode multiple columns in this space because we want the categorical values encoded but we do not care what the exact encoded values are. Any value can be attached to any numeric.

```python
df['Gender'] = encoder.fit_transform(df['Gender'])
df['Ever_Married'] = encoder.fit_transform(df['Ever_Married'])
df['Graduated'] = encoder.fit_transform(df['Graduated'])
df['Profession'] = encoder.fit_transform(df['Profession'])
df['Spending_Score'] = encoder.fit_transform(df['Spending_Score'])
df['Var_1'] = encoder.fit_transform(df['Var_1'])
```

For the column 'Segmentation I want the Letter for each cluster to encode numerical equivalent to the letter.

```python
df['Segmentation'] = df['Segmentation'].map({'A': 1, 'B': 2, 'C': 3, 'D': 4})
```

Finally, we save the new .csv file

```python
df.to_csv("Final_encoded.csv", index=False)
```
