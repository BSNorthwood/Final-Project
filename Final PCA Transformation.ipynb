In this space we will be running PCA.
We start by loading in all needed libraries.

```python
import pandas as pd
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
```

We load in the dataset.

```python
df = pd.read_csv("Final_encoded.csv")
```

For this we say to drop all non numeric columns

```python
df_numeric = df.select_dtypes(include=['number'])
```

We indicate how many PCA components we are looking for, this can be determined by looking at explained variance after the PCA has been ran.

```python
pca = PCA(n_components=2) 
```

WE now fit and transform the data

```python
principal_components = pca.fit_transform(df_numeric)
```

We create the DataFrame with the principal components

```python
pca_df = pd.DataFrame(data=principal_components, columns=['PC1', 'PC2'])
```

We save the outcome of the PCA

```python
pca_df.to_csv("Final_encoded_PCA.csv", index=False)
```

We can now plot the explained varianve into a visual with the code below.

```python
plt.figure(figsize=(8, 5))
plt.plot(range(1, len(pca.explained_variance_ratio_) + 1), pca.explained_variance_ratio_, marker='o')
plt.xlabel("Principal Component")
plt.ylabel("Explained Variance Ratio")
plt.title("Explained Variance of Principal Components")
plt.grid()
plt.show()
```

We want to check the explained variance.

```python
print(pca.explained_variance_ratio_) 
```

The results above mean that almost all vairance is captured with 1 principal componenet but we used 2 just in case.
