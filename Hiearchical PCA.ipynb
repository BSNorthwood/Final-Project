Here we will run the hierarchical modeling on the PCA dataset.
We load all needed Libraries

```python
import pandas as pd
import scipy.cluster.hierarchy as sch
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
```

We load the dataset

```python
df_pca = pd.read_csv("Final_encoded_PCA.csv")
```

We select the features we wish to use.

```python
df_selected = df_pca[['PC1', 'PC2']]
```

We standardize the data for this model

```python
scaler = StandardScaler()
df_scaled = scaler.fit_transform(df_selected)
```

We perform the hierarchical clustering

```python
linkage_matrix = sch.linkage(df_scaled, method='ward')  # 'ward' minimizes variance within clusters
```

We now plot and display the dendogram

```python
plt.figure(figsize=(10, 6))
sch.dendrogram(linkage_matrix, labels=df_pca.index, leaf_rotation=90, leaf_font_size=10)
plt.xlabel("Data Points")
plt.ylabel("Distance")
plt.title("Hierarchical Clustering Dendrogram (PCA)")
plt.grid()
plt.show()
```
