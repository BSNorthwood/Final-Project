Here we will run K-Means on the PCA Dataset
First we load in all libraries

```python
import pandas as pd
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
```

Now we load in the dataset 

```python
df = pd.read_csv("Final_encoded_PCA.csv")
```

Now we select the principal components for clustering.

```python
df_selected = df[['PC1', 'PC2']]
```

We define the amount of clusters we are looking for.

```python
num_clusters = 3
```

We initalize the K-Means modeling

```python
kmeans = KMeans(n_clusters=num_clusters, random_state=42)
df_selected['Cluster'] = kmeans.fit_predict(df_selected)
```

We save the the outcome to a new .csv

```python
df_selected.to_csv("Final_encoded_PCA_clustered.csv", index=False)
```

We now plot the clusters and create visualization

```python
plt.figure(figsize=(8, 6))
plt.scatter(df_selected['PC1'], df_selected['PC2'], c=df_selected['Cluster'], cmap='viridis', edgecolors='k')
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("K-Means Clustering on PCA-Transformed Data")
plt.colorbar(label="Cluster")
plt.grid()
plt.show()
```
