In this space we will be combing the train and test data to make the .csv file titled, 'Final_project_dataset.csv'
We wil first import libraries that we will use.

```python
import pandas as pd
```

Here we specify each file name.

```python
file1 = 'Test.csv'
file2 = 'Train.csv'
```

We now determine the output files name.

```python
output_file = 'Final_Project_dataset.csv'
```

We have to read each .csv file

```python
df1 = pd.read_csv(file1)
df2 = pd.read_csv(file2)
```

We now give the command to combine the files.

```python
combined_df = pd.concat([df1, df2], ignore_index=True)
```

We turn the combined files into the new .csv destination.

```python
combined_df.to_csv(output_file, index=False)
```
